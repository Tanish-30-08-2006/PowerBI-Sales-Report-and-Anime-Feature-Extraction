```python
import numpy as np
import pandas as pd
from IPython.display import display
```

```python
df1 = pd.DataFrame({
    'A': [1, 2, 3, 4, 5],
    'B': [10, 20, 30, 40, 50],
    'C': [100, 200, 300, 400, 500]
})
display(df1)
```

```python
#--------SQUARE EVERY ELEMENT IN FIRST COLUMN------------#

def square(x):
    square = x**2
    return square

squared_B_df1 = df1['B'].apply(square)

```

```python
# changes are not made yet so make the actual dataframe change..
display(df1)

df1['B'] = squared_B_df1 

display(df1)
```

```python
#-----------CREATE NEW COLUMN D WITH SQUARES OF ELEMENTS OF C---------#
def square(x):
    square = x**2
    return square
 
df1['D'] = df1['B'].apply(square)
display(df1['D'])
display(df1)
```
