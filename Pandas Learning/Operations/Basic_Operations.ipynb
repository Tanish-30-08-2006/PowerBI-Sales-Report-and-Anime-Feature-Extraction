```python
import numpy as np
import pandas as pd
from IPython.display import display
```

```python
df1 = pd.DataFrame({
    'A': [1, 2, 3, 4, 5],
    'B': [10, 20, 30, 40, 50],
    'C': [100, 200, 300, 400, 500]
})
display(df1)
```

```python
#----------- ADD 10 IN FIRST COLUMN [ 'A' ]-----------#

display(df1['A']+10)
display(df1)
```

```python
#---------MULTIPLY BY 2 IN SECOND COLUMN-------------#

display(df1['B']*2)
display(df1)  # still stays the same as above line creates a copy and does not touch the original array 
```

```python
# -------------ADD 20 IN FIRST TWO COLUMNS----------#

display(df1)
display(df1[['A','B']] + 20)
display(df1)
```

```python
# ----------ADD 10 IN FIRST ROW-------------#

display(df1.loc[[1]]+10)
```

```python
# ADD 20 IN FIRST AND SECOND ROW 

# index 0 and 1 --> 20 add
display(df1.loc[[0,1]]+20)
```
