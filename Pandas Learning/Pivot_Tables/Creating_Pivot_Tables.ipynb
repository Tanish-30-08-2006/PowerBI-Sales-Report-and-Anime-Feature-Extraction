```python
import pandas as pd
import numpy as np
from IPython.display import display
```

```python
data = {
    'Date': pd.date_range('2023-01-01', periods=20),
    'Product': ['A', 'B', 'C', 'D'] * 5,
    'Region': ['East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West',
               'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    'Sales': np.random.randint(100, 1000, 20),
    'Units': np.random.randint(10, 100, 20),
    'Rep': ['John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary',
            'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice']
}

df = pd.DataFrame(data)

df['Month'] = df['Date'].dt.month_name()
df['Quarter'] = 'Q' + df['Date'].dt.quarter.astype(str)

display(df)
```

```python
#---------------TO SEE REGION WISE MEAN SALES OF ALL PRODUCTS------------------------------------#

# here we want sales so values = sales , and indexes as regions as we want sales region wise ,
# and columns = products as we want region wise products as columns and sales must be shwon

pivot_table = pd.pivot_table( df, values='Sales', index='Region' , columns = 'Product') 

display(pivot_table)
# values , index and columns must be filled .... to use pivot table  

# By default , pivot tables calculates -----MEAN----- as aggfunc ---> aggregation function..

```

```python
#----------------TO CALCULATE SUM , STD DEVIATION ETC AS AGGREGATION FUNCTION-------------#

#----------SUM----------#

# TO CALCULATE SUM OF SALES(VALUES) GROUPED BY REGION(INDEXES) FOR EVERY PRODUCT (COLUMNS)
pivot_table = pd.pivot_table( df , values='Sales' , index = 'Region' , columns = 'Product', aggfunc='sum',margins='True')
display(pivot_table)
```

```python
#-------------STD DEV---------#

# TO CALCULATE STD DEV OF SALES(VALUES) GROUPED BY REGION(INDEXES) I.E REGIONWISE FO EVERY PRODUCT(COLUMNS)
pivot_table = pd.pivot_table( df ,values = 'Sales' , index = 'Region' , columns = 'Product' , aggfunc='std',margins='True')

display(pivot_table)

# margins = true gives sum of all rows and columns as a new row and column at the end 
```

```python
#------VARIANCE--------#

# TO CALCULATE VARIANCE OF SALES(VALUES) GROUPED BY REGION(INDEXES) FOR ALL PRODUCTS(COLUMNS)

pivot_table = pd.pivot_table(df , values = 'Sales' , index = 'Region' , columns = 'Product',aggfunc='var',margins='True')
display(pivot_table)
```

```python
# USECASES OF PIVOT TABLES MAINLY IN  ----> CLUSTERED HEAT MAP CREATION IN SEABORN AND MATPLOTLIB
```

```python
#---------PIVOT TABLES FOR MULTIPLE VALUES ---------------#

# TO CALCULATE MEAN OF SALES AND UNITS (VALUES) GROUPED BY REGION(INDEXES) FOR ALL PRODUCTS(COLUMNS)

pivot_table = pd.pivot_table(df , values = ['Sales' , 'Units'],index = 'Region', columns ='Product' ,aggfunc = 'mean',margins='True')
display(pivot_table)
```
