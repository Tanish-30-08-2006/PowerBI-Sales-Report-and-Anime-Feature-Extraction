```python
import pandas as pd
import numpy as np
from IPython.display import display
```

```python
# Show all columns without truncation
pd.set_option('display.max_columns', None)

# Limit the number of rows shown (prevents a wall of text)
pd.set_option('display.max_rows', 100)

# Set the max width of a column (prevents long text from stretching the table)
pd.set_option('display.max_colwidth', 100)

# Reduce the decimal places for cleaner numbers
pd.set_option('display.precision', 2)
```

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
df = pd.DataFrame(data)
display(df)
```

```python
# Group by category and calculate  : SUM , AVERAGE , STD DEVIATION , VARIANCE --> SALES

grp_cat  = df.groupby('Category')
display(grp_cat)
print("\n")
# we cant view group by , output will be : new object created at some address

for cat1,cat2 in grp_cat:
    display(cat1)
    display(cat2)
```

```python
#-----------CALCULATE SUM , AVG , MEAN , VARIANCE...
grp_cat = df.groupby('Category')

display(grp_cat['Sales'].sum())

print("\nMean: ")
display(grp_cat['Sales'].mean())

print("\nStandard Deviation: ")
display(grp_cat['Sales'].std())

print("\nVariance: ")
display(grp_cat['Sales'].var())
```

```python
#---------------GROUP BY STORES AND :  SUM , AVG --> SALES-------------#

grp_stores = df.groupby('Store')
for s1,s2 in grp_stores : 
    display(s1)
    display(s2)
```

```python
#-----------SUM , AVG ----------#

display(grp_stores['Sales'].sum())

print("\n")
display(grp_stores['Sales'].mean())
```

```python
#----------GROUP BY STORE AND CATEGORY : --> SALES SUM -------------#

# TWO TYPES OF DATA : QUALITATIVE AND QUANTATIVE DATA 
# QUALITATIVE DATA : CATEGORICAL DATA  ( Stores ,  category , male female ....)
# QUANTATIVE DATA :  (quant_in_stock , date of sale , sales )

# We use group by for categorical data..


grp_cat_store = df.groupby(['Category', 'Store'])
display(grp_cat_store['Sales'].sum())

print("\n")
grp_store_cat = df.groupby(['Store', 'Category'])
display(grp_store_cat['Sales'].sum())

```
