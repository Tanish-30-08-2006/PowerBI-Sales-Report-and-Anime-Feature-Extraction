```python
import pandas as pd
import numpy as np
from IPython.display import display
```

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
df = pd.DataFrame(data)
display(df)
```

```python
#------------ALL SALES STATISTICAL OPERATIONS---------------#

sales = df['Sales']
display(sales)

sales.agg(['sum' , 'mean' , 'min' , 'max' , 'count' , 'std' , 'median','var'])
```

```python
# --- ALL STATISTICAL OPERATIONS ON SALES GROUPED BY CATEGROY-------#

grp_by_cat = df.groupby('Category')
for cat1 , cat2 in grp_by_cat:
    display(cat1 ,cat2 )

sales_grp_by_cat = grp_by_cat['Sales']
for cat1 , cat2 in sales_grp_by_cat:
    display(cat1,cat2)

sales_grp_by_cat.agg(['sum' , 'mean' , 'min' , 'max' , 'count' , 'std' , 'median','var'])
```

```python
# --- ALL STATISTICAL OPERATIONS ON SALES GROUPED BY CATEGORY AND STORES---------#

grp_by_cat_store = df.groupby(['Category','Store'])
for cat1,cat2 in grp_by_cat_store:
    display(cat1,cat2)

```

```python
sales_grp_by_cat_store = grp_by_cat_store['Sales']

for cat1,cat2 in sales_grp_by_cat_store:
    display(cat1,cat2)
```

```python
display(sales_grp_by_cat_store.agg(['sum' , 'mean' , 'min' , 'max' , 'count' , 'std' , 'median','var']))
```
