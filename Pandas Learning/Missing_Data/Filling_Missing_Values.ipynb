```python
import pandas as pd
import numpy as np

data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}

df = pd.DataFrame(data)
display(df)
df = pd.DataFrame(data.values(), index=data.keys()).T
display(df)
```

```python
 #----------------FILLING MISSING VALUES ACROSS WHOLE DATAFRAME----------------#

display(df.fillna(value = 0))

```

```python
#---------------FILLING MISSING VALUES COLUMN WISE----------------------#

values_to_fill = {
    'A' : 1000,
    'B' : 2000,
    'C' : 3000,
    'D' : 4000
}

# REMEMBER WHILE FILLING MISSING VALUES  COLUMN WISE 
# WE USE DICTIONARY WITH KEYS AS COLUMNS AND VALUES AS REPLACEMENTS OF NULL IN THAT COLUMN..
display(df.fillna(value = values_to_fill))

display(df) # still the same ..

values_to_fill = {
    'A' : 'a',
    'B' : 'b',
    'C' : 'c',
    'D' : 'd'
}
display(df.fillna(value = values_to_fill))
```
