```python
import pandas as pd
import numpy as np

data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}

df = pd.DataFrame(data)
display(df)
df = pd.DataFrame(data.values(), index=data.keys()).T
display(df)
```

```python
#--------------TO VIEW ROWS WITHOUT MISSING DATA-----------#

display(df.dropna()) # all rows without missing values # by default thresh = 4 
```

```python
#-----------------ADDING THRESHOLS WITH DROPPING ROWS-------------#

display(df.dropna(thresh=3)) # dropna measn dropping null so accessing not null values 

display(df.dropna(thresh=1)) # all rows which have atleast one non null values.....

display(df.dropna(thresh=4)) 

display(df.dropna(thresh=5)) 

# thresh = 3 atleast 3 not null values as dropna means dropping null values 
```
