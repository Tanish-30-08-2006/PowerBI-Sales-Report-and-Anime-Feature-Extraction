```python
import pandas as pd
import numpy as np

data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}

df = pd.DataFrame(data)
display(df)
df = pd.DataFrame(data.values(), index=data.keys()).T
display(df)
```

```python
#--------------------TO FIND NULL VALUES AND ITS SUM ACROSS WHOLE DATA(BOOL)-------------------#

bool_isna = df.isna()
display(bool_isna)

df.isna().sum() # gives how many null values in each column 

# We will have 100 Columns so to find columns which have some missing values or not 
# meaning to find the columns which have atleast one missing value 

df.isna().any() # for every column true or false for presence of missing values 
```
