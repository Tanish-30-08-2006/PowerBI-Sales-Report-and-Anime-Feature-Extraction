```python
import pandas as pd
import numpy as np
from IPython.display import display
```

```python

data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
columns = ["Name" , "Age" , "City" , "Salary"]
df2 = pd.DataFrame(data_list)
display(df2)
df2 = pd.DataFrame(data_list , columns = columns)
display(df2)
```

```python
#------------------SELECTING  SINGLE ROW-----------------#

display(df2.loc[0])
print("\n")
display(df2.loc[2])
```

```python
# ----------------SELECTING MULTIPLE ROWS---------------------#

display(df2)

display(df2.loc[[0,1]]) # for multiple rows viewing we pass list of lists of row labels or indexes of that rows 
display(df2.loc[[1,2,3]])

```

```python
#----------VIEWING SPECIFIC COLUMNS OF SPECIFIC ROWS---------------#

display(df2)

display(df2.loc[[1,2,3]]['Name'])

display(df2.loc[[1,2]][['City' , 'Salary']])
```
