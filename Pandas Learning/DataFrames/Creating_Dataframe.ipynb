```python
import numpy as np
import pandas as pd
```

```python
#---------------CREATING DATA FRAME USING DICTIONARY ---------------#

data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}
df1 = pd.DataFrame(data)
print(df1)
# Name , Age , city , salary were series and combined to make a series 
```

```python
#------------------CREATING DATAFRAME USING LISTS------------------#

data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
df2 = pd.DataFrame(data_list)
print(df2)
```

```python
#--------TO CHANGE COLUMNS NAME OR ROWS NAME WHEN USING LISTS SO DEFAULT LABELLING -----------#


columns = ["Name" , "Age" , "City" , "Salary"]
df2 = pd.DataFrame(data_list , columns=columns)
print(df2)
```
