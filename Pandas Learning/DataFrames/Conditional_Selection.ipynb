```python
import pandas as pd
import numpy as np
from IPython.display import display
```

```python
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
columns = ["Name" , "Age" , "City" , "Salary"]
df2 = pd.DataFrame(data_list)
display(df2)
df2 = pd.DataFrame(data_list , columns = columns) # every tuple as a list 
display(df2)
```

```python
#----------------CONDITIONAL SELECTION----------------#

# SELECT PEOPLE AGED ABOVE 30...

bool_age = df2['Age']>30
display(bool_age)
df2[bool_age]
```
