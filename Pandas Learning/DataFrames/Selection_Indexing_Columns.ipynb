```python
import pandas as pd
import numpy as np
```

```python
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
columns = ["Name" , "Age" , "City" , "Salary"]
df2 = pd.DataFrame(data_list , columns=columns)
print(df2)
```

```python
#----------TO SELECT SINGLE COLUMNS----------------#

# we can't use df2[0] as we need to access by labels as we changes it from 0 1..  to Name Age ..
print(df2['Name'])
print("\n")
print(df2['Age'])
print("\n")
print(df2['City'])
print("\n")
```

```python
#----------TO SELECT MULTIPLE COLUMNS----------------#

# we need lists of lists of labels of columns to select multiple tables 
print(df2[['Name' , 'City']])
```
