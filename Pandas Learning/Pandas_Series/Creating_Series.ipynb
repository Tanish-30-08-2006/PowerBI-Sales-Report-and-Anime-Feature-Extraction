```python
import numpy as np
import pandas as pd
```

```python
#--------------SERIES DEFINITION------------------#

# A series is a 1D labelled array capable of holding any datatype. The axis labels are collectively called index

# Arrays are visualized horizontally but if i visualize data in vertical i.e stacking data in vertical is called SERIES

# There are multiple ways to create series in pandas
```

```python
#----------------SERIES CREATION USING LISTS-------------------#

# first lets create labels (indexes for series) 


my_list = [10,20,30]


# Now to create a data series we need data points (my_list) and index points which are called labels

# ----------default index points(labels)----------#

print("Default labels : ")
series1 = pd.Series(my_list)
print(series1)

# we can see default labels are starting from 0 vertcially 

# --------------custom labels---------------#

labels = ['a','b','c'] # we want index as a b c ..
series1 = pd.Series(my_list, index=labels)
print("\nCustom labels created : \n",series1)

```

```python
#----------------SERIES CREATION USING ARRAYS-------------------#

arr = np.linspace(1,7,3) # floating points are datatype for linspace as evenly spacing oftenly goes to decimals...
series1  = pd.Series(arr,index=None)
print("Using series  : ")
print(series1)

# ---------using custom labels------#

labels  = ['l' , 'm' , 'n']
series1 = pd.Series(arr,index=labels)
print("Using custom labels : ")
print(series1)

```

```python
#----------------SERIES CREATION USING DICTIONARY-------------------#


# dict has bydefault keys and values or labels and datapoints 

dicty = {1:10 , 2:20 , 3:30}
series1 = pd.Series(dicty.values() , index=dicty.keys())
print(series1)
series2 = pd.Series(dicty)
print(series2)
```
