```python
import numpy as np
import pandas as pd
from IPython.display import display
```

```python

df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})

df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})

display(df1 , df2)
```

```python
#------------------CONCATINATION------------------#

display(pd.concat([df1,df2]))

display(pd.concat([df2,df1]))

# common columns are stacked........
```
