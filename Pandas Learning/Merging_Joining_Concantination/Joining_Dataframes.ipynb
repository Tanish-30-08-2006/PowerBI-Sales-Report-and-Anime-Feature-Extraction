```python
import numpy as np
import pandas as pd
from IPython.display import display
```

```python
# First Dataframe
df1 = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie']
}, index=[1, 2, 3])

# Second DataFrame
df2 = pd.DataFrame({
    'score': [85, 90, 75]
}, index=[2, 3, 4])

display(df1 , df2)
```

```python
#------------------JOIN DATAFRAMES------------------#

# pd.merge(df1,df2) wont work as there is no common attrbute (column )on which we can join


display(df1.join(df2,how='outer'))

display(df1.join(df2,how='left'))

display(df1.join(df2,how='right'))

```

```python
# SIIMILARY , 

display(df1 , df2)

display(df2.join(df1,how='outer'))

display(df2.join(df1,how='left'))

display(df2.join(df1,how='right'))
```
